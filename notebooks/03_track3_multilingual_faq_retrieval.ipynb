# Track 3 starter: Multilingual FAQ retrieval
A tiny retrieval baseline over the FAQ file (English + draft Nigerian Pidgin). Hausa, Yoruba and Igbo columns are intentionally empty: **finding and validating good translations with native speakers is part of the challenge.**

```python
import sys; sys.path.append('../src')
import pandas as pd, numpy as np
from ai4elections import load_csv, load_jsonl
pd.set_option('display.max_columns', 30)
```

```python
faq = load_csv('voter_info_faq_multilingual.csv')
faq[['faq_id','topic','en','pcm_draft']].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

corpus = pd.concat([faq['en'], faq['pcm_draft']], ignore_index=True)
ids = list(faq['faq_id'])*2
vec = TfidfVectorizer(analyzer='char_wb', ngram_range=(2,4)).fit(corpus)
M = vec.transform(corpus)

def ask(q, k=3):
    sims = cosine_similarity(vec.transform([q]), M).ravel()
    top = sims.argsort()[::-1][:k]
    return [(ids[i], corpus[i], round(float(sims[i]),2)) for i in top]

ask('I no get PVC, I fit vote?')
```
