```python
sv = load_csv('accessibility_barriers_survey_synthetic.csv')
# k-anonymity check on quasi-identifiers
qi = ['state','age_band','gender','disability_type']
k = sv.groupby(qi).size()
print('Groups with k<3:', (k<3).sum(), 'of', len(k))
```

```python
# Suppress small cells before publishing aggregates
def suppress(df, cols, min_k=5):
    g = df.groupby(cols).size().reset_index(name='n')
    g.loc[g['n']<min_k,'n'] = np.nan
    return g
suppress(sv, ['state','disability_type']).head()
```

```python
# Toy log-anomaly monitor: bursts of incident reports per hour
inc = load_csv('incident_reports_synthetic.csv'); inc['hour'] = pd.to_datetime(inc['timestamp']).dt.floor('h')
counts = inc.groupby('hour').size(); z = (counts-counts.mean())/counts.std()
z[z>2]
```
